## Bonus — cross-device robustness (emulator vs real)

```python
import warnings; warnings.filterwarnings("ignore")
import pandas as pd
import numpy as np

# --- load both aggregate tables ---
real = pd.read_csv("results/krono_real/krono_real_aggregate_ci.csv")
emu  = pd.read_csv("results/krono_emu/krono_emu_aggregate_ci.csv")
real["device"] = "real"
emu["device"]  = "emu"

# metrics worth comparing across devices
METRICS = ["macro_f1", "stab_adjjac", "seed_adjjac", "del_median",
           "shap_lime_adjjac_disc1", "intr_shap_adjjac"]

# --- side-by-side table: same (modality, model) row, emu vs real ---
both = pd.concat([real, emu], ignore_index=True)
comp = both.pivot_table(index=["modality", "model"], columns="device",
                        values=METRICS)
# flatten column names and compute the emu-vs-real gap per metric
comp.columns = [f"{m}_{d}" for m, d in comp.columns]
for m in METRICS:
    if f"{m}_real" in comp and f"{m}_emu" in comp:
        comp[f"{m}_gap"] = (comp[f"{m}_real"] - comp[f"{m}_emu"]).round(3)

pd.set_option("display.width", 200, "display.max_columns", 50)
print("=== Emulator vs Real: per (modality, model) ===")
print(comp[[f"{m}_real" for m in METRICS] + [f"{m}_emu" for m in METRICS]].round(3).to_string())

print("\n=== Emu-vs-Real gaps (real minus emu) ===")
print(comp[[f"{m}_gap" for m in METRICS if f"{m}_gap" in comp]].to_string())

comp.to_csv("results/krono_device_comparison.csv")
print("\nsaved results/krono_device_comparison.csv")
```

## Contribution 4 — temporal / concept drift (KronoDroid-only capability)

```python
import pandas as pd
from ream.core.kronodroid import load_kronodroid_with_year
from ream.contributions import c4_temporal as c4

mods, y, year = load_kronodroid_with_year(device='emu')
# run temporal-vs-random for each modality at a chosen cutoff year
for tag, X in mods.items():
    df = X.copy(); df['label']=y.values; df['year']=year.values
    df = df.dropna(subset=['year'])
    feat=[c for c in df.columns if c not in ('label','year')]
    res = c4.run_temporal(df, feat, 'label', 'year', cutoff=2016, model_kind='lgbm', seed=0)
    print(tag, 'stability drop under drift:', round(res['stability_drop'],3),
          '| faithfulness change:', round(res['faithfulness_change'],3))
```
