# 03 Contribution 3 — Adversarial explanation robustness
Prediction-preserving attacks that destabilize explanations.

```python
import sys; sys.path.insert(0,'..')
import numpy as np
from ream.core import data as D
from ream.core.models import train_model
from ream.core.metrics import detect_feature_type, common_correct_indices
from ream.contributions import c3_adversarial as c3
```

```python
import warnings
warnings.filterwarnings("ignore") 
```

```python
# run attack on one (dataset,modality,model)
X,y,cn=D.load_modality('maldroid','M2')
Xtr,Xte,ytr,yte=D.make_split(X,y,42)
clf=train_model('lgbm',Xtr,ytr,seed=42)
mask={'m':(clf.predict(Xte.to_numpy())==yte.to_numpy())}
common=common_correct_indices(mask)[:100]
isb=detect_feature_type(Xtr.to_numpy())
xs=[Xte.to_numpy()[i] for i in common]; cs=[int(yte.to_numpy()[i]) for i in common]
res=c3.run_attack(clf,'lgbm',Xtr.to_numpy(),xs,cs,isb,X.shape[1],budget=40,seed=1)
print(res)
```
